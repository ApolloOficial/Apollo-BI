#### Ler o .env

```python
import os
from dotenv import load_dotenv, find_dotenv

load_dotenv(find_dotenv(usecwd=True))

host = os.getenv("NEON_HOST")
banco = os.getenv("NEON_DB")
usuario = os.getenv("NEON_USER")
senha = os.getenv("NEON_PASSWORD")
```

#### Conectar com o Banco de Dados

```python
df = (spark.read.format("jdbc")
      .option("url", f"jdbc:postgresql://{host}:5432/{banco}?sslmode=require")
      .option("user", usuario)
      .option("password", senha)
      .option("driver", "org.postgresql.Driver")
      .option("dbtable", "dm.dim_company")
      .load())
```

#### Testar a conexão

```python
display(df)
```
